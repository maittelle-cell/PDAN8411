```python
###STEP 9

# STEP 9.1: Prepare the customer review text

# Combine the review title and full customer comment
df['review_text'] = (
    df['Review title'].fillna('').astype(str) + ' ' +
    df['Full comment / indexed extract'].fillna('').astype(str)
)

# Display the first five combined reviews
print(df[['Review title', 'review_text']].head())
```

```python
# STEP 9.2

# Import regular expressions and stop words
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Create a function to clean customer review text
def clean_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove numbers and punctuation
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Split text into individual words
    words = text.split()

    # Remove common English stop words
    words = [
        word for word in words
        if word not in ENGLISH_STOP_WORDS
    ]

    # Join the cleaned words
    return ' '.join(words)


# Apply the cleaning function to all reviews
df['cleaned_review'] = df['review_text'].apply(clean_text)

# Display original and cleaned reviews
print(df[['review_text', 'cleaned_review']].head())
```

```python
#STEP 9.3
# Check whether any cleaned reviews are empty
empty_reviews = (df['cleaned_review'].str.strip() == '').sum()

print("Number of empty cleaned reviews:", empty_reviews)

# Display the length of each cleaned review
df['review_length'] = df['cleaned_review'].apply(
    lambda x: len(x.split())
)

print("\nReview length summary:")
print(df['review_length'].describe())
```

```python
#STEP 10

#STEP 10.1: Identify the most frequently used words

from collections import Counter

# Combine all cleaned customer reviews
all_words = ' '.join(df['cleaned_review']).split()

# Count how frequently each word appears
word_frequency = Counter(all_words)

# Display the 20 most common words
most_common_words = word_frequency.most_common(20)

print("TOP 20 MOST FREQUENT WORDS")
print("=" * 40)

for word, frequency in most_common_words:
    print(word, ":", frequency)
```

```python
#STEP 10.3
# Visualise the 20 most frequently used words

word_df = pd.DataFrame(
    most_common_words,
    columns=['Word', 'Frequency']
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=word_df,
    x='Frequency',
    y='Word'
)

plt.title('Most Frequently Used Words in Bonitas Reviews')
plt.xlabel('Word Frequency')
plt.ylabel('Words')

plt.tight_layout()
plt.show()
```

```python

###STEP 11: Prepare data for Latent Dirichlet Allocation

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Convert customer reviews into word-count features
count_vectorizer = CountVectorizer(
    max_df=1.0,
    min_df=2,
    max_features=500
)

# Create the document-term matrix
document_matrix = count_vectorizer.fit_transform(
    df['cleaned_review']
)

print("Document-term matrix shape:")
print(document_matrix.shape)
```

```python
#STEP 11.2
# Train an LDA topic modelling model

lda_model = LatentDirichletAllocation(
    n_components=3,
    random_state=42,
    max_iter=20,
    learning_method='batch'
)

# Fit the model to the customer reviews
lda_model.fit(document_matrix)

print("LDA topic modelling completed.")
```

```python
#STEP 11.3

# Display the most important words in each topic

feature_names = count_vectorizer.get_feature_names_out()

for topic_number, topic in enumerate(lda_model.components_):

    print("\nTOPIC", topic_number + 1)
    print("-" * 40)

    top_word_indices = topic.argsort()[-10:][::-1]

    top_words = [
        feature_names[index]
        for index in top_word_indices
    ]

    print(", ".join(top_words))
```

```python
# STEP 12.2: CALCULATE SENTIMENT POLARITY SCORES

# Use the original review text for sentiment analysis
df['review_text'] = df['review_text'].fillna('').astype(str)

# Calculate sentiment polarity
df['sentiment_score'] = df['review_text'].apply(
    lambda text: TextBlob(text).sentiment.polarity
)

# Display the results
print("SENTIMENT POLARITY SCORES")
print("=" * 50)

print(df[['Review ID', 'review_text', 'sentiment_score']].head(10))
```

```python
# STEP 12.3: CLASSIFY SENTIMENT

def classify_sentiment(score):

    if score < -0.05:
        return 'Negative'

    elif score > 0.05:
        return 'Positive'

    else:
        return 'Neutral'


# Apply sentiment classification
df['sentiment'] = df['sentiment_score'].apply(classify_sentiment)

# Display the classified reviews
print("CUSTOMER SENTIMENT CLASSIFICATION")
print("=" * 50)

print(df[['Review ID', 'Star rating', 'sentiment_score', 'sentiment']].head(15))
```

```python
# STEP 12.5: SENTIMENT DISTRIBUTION CHART

sentiment_order = ['Negative', 'Neutral', 'Positive']

sentiment_counts = (
    df['sentiment']
    .value_counts()
    .reindex(sentiment_order, fill_value=0)
)

plt.figure(figsize=(9, 6))

ax = sns.countplot(
    data=df,
    x='sentiment',
    order=sentiment_order
)

# Add review counts and percentages
for i, category in enumerate(sentiment_order):

    count = sentiment_counts[category]
    percentage = (count / len(df)) * 100

    ax.text(
        i,
        count + 0.2,
        f'{count} ({percentage:.1f}%)',
        ha='center'
    )

plt.title('Sentiment Distribution of Bonitas HelloPeter Reviews')
plt.xlabel('Sentiment Category')
plt.ylabel('Number of Reviews')

plt.tight_layout()
plt.show()
```

The analysis provides an indication of the emotional tone expressed in the customer reviews. 
Comparing sentiment categories with star ratings also helps identify whether the language used by customers corresponds with their numerical ratings.

```python
#STEP 13

# STEP 13.1: Extract numerical features using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

# Initialise the TF-IDF vectoriser
tfidf = TfidfVectorizer(
    max_features=1000,
    ngram_range=(1, 2),
    min_df=1
)

# Convert cleaned reviews into numerical features
X = tfidf.fit_transform(df['cleaned_review'])

# Display the shape of the feature matrix
print("TF-IDF feature matrix shape:")
print(X.shape)

print("\nNumber of customer reviews:", X.shape[0])
print("Number of extracted features:", X.shape[1])
```

```python
#STEP 13.2

# Display the vocabulary extracted by TF-IDF

feature_names = tfidf.get_feature_names_out()

print("FIRST 30 EXTRACTED FEATURES")
print("=" * 40)

print(feature_names[:30])
```

```python
#STEP 14 

# STEP 14: Prepare data for machine learning

# X represents the customer review text
X_text = df['cleaned_review']

# y represents the sentiment category
y = df['sentiment']

# Display the number of observations in each class
print("SENTIMENT CLASS COUNTS")
print(y.value_counts())

# Check the number of unique sentiment classes
print("\nNumber of sentiment classes:", y.nunique())
```

```python
#STEP 14.2
# Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("TRAINING DATA:", len(X_train))
print("TESTING DATA:", len(X_test))

print("\nTraining sentiment distribution:")
print(y_train.value_counts())

print("\nTesting sentiment distribution:")
print(y_test.value_counts())
```

```python
#STEP 15

# STEP 15: Import machine-learning algorithms

from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectPercentile, chi2
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression

# Create a Naive Bayes pipeline
nb_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=1000,
        ngram_range=(1, 2)
    )),
    ('feature_selection', SelectPercentile(
        score_func=chi2,
        percentile=50
    )),
    ('classifier', MultinomialNB())
])

# Create a Logistic Regression pipeline
lr_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=1000,
        ngram_range=(1, 2)
    )),
    ('feature_selection', SelectPercentile(
        score_func=chi2,
        percentile=50
    )),
    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

print("Both machine-learning pipelines have been created.")
```

```python
#STEP 15.2

# Train the Naive Bayes classification model

nb_pipeline.fit(X_train, y_train)

# Predict sentiment for the testing data
nb_predictions = nb_pipeline.predict(X_test)

print("Naive Bayes model training completed.")
```

```python
#STEP 15.3
# Train the Logistic Regression model

lr_pipeline.fit(X_train, y_train)

# Predict sentiment for the testing data
lr_predictions = lr_pipeline.predict(X_test)

print("Logistic Regression model training completed.")
```

```python
#STEP 14

# STEP 16.1: Compare model accuracy

from sklearn.metrics import accuracy_score

# Calculate Naive Bayes accuracy
nb_accuracy = accuracy_score(
    y_test,
    nb_predictions
)

# Calculate Logistic Regression accuracy
lr_accuracy = accuracy_score(
    y_test,
    lr_predictions
)

# Display model comparison
model_comparison = pd.DataFrame({
    'Model': [
        'Multinomial Naive Bayes',
        'Logistic Regression'
    ],
    'Accuracy': [
        nb_accuracy,
        lr_accuracy
    ]
})

print(model_comparison)
```

```python
#STEP 16.2
# Display detailed performance metrics

from sklearn.metrics import classification_report

print("NAIVE BAYES CLASSIFICATION REPORT")
print("=" * 50)

print(classification_report(
    y_test,
    nb_predictions,
    zero_division=0
))

print("\nLOGISTIC REGRESSION CLASSIFICATION REPORT")
print("=" * 50)

print(classification_report(
    y_test,
    lr_predictions,
    zero_division=0
))
```

```python
#STEP 16.3
# Compare actual and predicted sentiment using a confusion matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Generate confusion matrix for Logistic Regression
cm = confusion_matrix(
    y_test,
    lr_predictions,
    labels=['Negative', 'Neutral', 'Positive']
)

# Display confusion matrix
display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Negative', 'Neutral', 'Positive']
)

display_matrix.plot(cmap='Blues')

plt.title('Logistic Regression Confusion Matrix')

plt.tight_layout()
plt.show()
```

```python
#STEP 16.3
# Visualise the accuracy of both models

plt.figure(figsize=(9, 5))

sns.barplot(
    data=model_comparison,
    x='Model',
    y='Accuracy'
)

plt.title('Comparison of Sentiment Classification Models')
plt.xlabel('Machine Learning Model')
plt.ylabel('Accuracy')

plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```

```python
##STEP 17

# STEP 17.1: Improve Logistic Regression using GridSearchCV

from sklearn.model_selection import GridSearchCV, StratifiedKFold

# Define the parameter combinations to test
parameter_grid = {
    'tfidf__max_features': [500, 1000],
    'tfidf__ngram_range': [(1, 1), (1, 2)],
    'classifier__C': [0.1, 1, 10]
}

# Use two-fold cross-validation because the dataset is small
cross_validation = StratifiedKFold(
    n_splits=2,
    shuffle=True,
    random_state=42
)

# Create GridSearchCV
grid_search = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=parameter_grid,
    scoring='f1_macro',
    cv=cross_validation,
    n_jobs=-1
)

# Train and search for the best parameters
grid_search.fit(X_train, y_train)

print("BEST PARAMETERS")
print(grid_search.best_params_)

print("\nBEST CROSS-VALIDATION MACRO F1 SCORE")
print(grid_search.best_score_)
```

```python
#STEP 17.2

# Evaluate the improved Logistic Regression model

# Select the best model
best_model = grid_search.best_estimator_

# Predict testing sentiment
improved_predictions = best_model.predict(X_test)

# Calculate improved accuracy
improved_accuracy = accuracy_score(
    y_test,
    improved_predictions
)

print("IMPROVED MODEL ACCURACY:")
print(round(improved_accuracy, 4))

print("\nIMPROVED MODEL CLASSIFICATION REPORT")

print(classification_report(
    y_test,
    improved_predictions,
    zero_division=0
))
```

```python
#STEP 17.3

# Compare the original and improved Logistic Regression models

original_accuracy = accuracy_score(
    y_test,
    lr_predictions
)

improved_accuracy = accuracy_score(
    y_test,
    improved_predictions
)

improvement_comparison = pd.DataFrame({
    'Model': [
        'Original Logistic Regression',
        'Improved Logistic Regression'
    ],
    'Accuracy': [
        original_accuracy,
        improved_accuracy
    ]
})

print(improvement_comparison)

# Visualise the comparison
plt.figure(figsize=(9, 5))

sns.barplot(
    data=improvement_comparison,
    x='Model',
    y='Accuracy'
)

plt.title('Original vs Improved Model Performance')
plt.xlabel('Model')
plt.ylabel('Accuracy')

plt.ylim(0, 1)

plt.xticks(rotation=10)

plt.tight_layout()
plt.show()
```

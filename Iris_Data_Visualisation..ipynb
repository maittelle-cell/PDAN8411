```python
#load Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import seaborn as sns
```

```python
#load Datasets
iris_datasets=load_iris()
print("keys of iris_dataset:/n{}".format(iris_datasets.keys()))
print(iris_datasets.keys())
print(iris_datasets.target_names)
print(iris_datasets.feature_names)
print(iris_datasets.DESCR)
print(iris_datasets.data.shape)
print(iris_datasets.data[:5])
```

```python
#convert  to Dataframe
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns= iris.feature_names)
iris_df['species'] = iris.target

sns.pairplot(iris_df, hue="species")
plt.show()
```

```python

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

mean_values = df.mean(numeric_only=True)

print(mean_values)

mean_values.plot(kind='bar')

plt.title("Average Measurements of Iris Flowers")
plt.xlabel("Features")
plt.ylabel("Average Measurement (cm)")


plt.show()
```

```python

iris_dataset = load_iris()

iris_dataframe = pd.DataFrame(
    iris_dataset['data'],
    columns=iris_dataset['feature_names']
)

pd.plotting.scatter_matrix(
    iris_dataframe,
    figsize=(10, 10),
    c=iris_dataset['target']
)

plt.show()
```

```python
#Create a Histogram
plt.hist(df["sepal length (cm)"])

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")

plt.show()
```

```python

#Average Measurements Chart The greatest average measurement is  petal length The lowest average measurementis petal width.
#Scatter Plot 1 Shows an upward slope With greater increase in length of sepals, will come greater increase in length of petals. 
#The dots suggest there is an positive association between them.
#Majority of flowers have sepal lengths of 5-6cm,shorten and longest are the least common.
```
